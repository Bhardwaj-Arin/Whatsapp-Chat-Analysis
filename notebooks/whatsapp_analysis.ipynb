## Pre-Processing
 Our chats are in text file so we need to convert this text file into pandas dataframe
 

```python
import re # re stands for Regular Expressions.
import pandas as pd
```

```python
file_path = "../datasets/tagore_group.txt"

f = open(file_path, "r", encoding="utf-8")
```

open(file, mode, encoding) <br>
"r" :- Read the file <br>
"w" :- Write <br>
"a" :- Append <br>
encoding="utf-8" :- This tells Python how the text is encoded. Without it, emojis or special characters may not load correctly.

```python
data = f.read()
```

```python
print(data)
```

```python
print(repr(data[:300]))
```

```python
# Replace Unicode narrow no-break space with a normal space
data = data.replace('\u202f', ' ')
```

```python
print(repr(data[:150]))
```

```python
pattern = r'\d{1,2}/\d{1,2}/\d{2},\s\d{1,2}:\d{2}\s(?:am|pm)\s-\s'
```

```python
messages = re.split(pattern, data)[1:]
messages
```

```python
dates = re.findall(pattern, data)
dates
```

```python
df = pd.DataFrame({'message_date': dates, 'user_message': messages})
df['message_date'] = (df['message_date'].str.replace('am', 'AM', regex=False).str.replace('pm', 'PM', regex=False))
df['message_date'] = pd.to_datetime(df['message_date'], format='%d/%m/%y, %I:%M %p - ')
df = df[['user_message', 'message_date']]
df.rename(columns={'message_date':'date'}, inplace=True)
df.head()
```

```python
print(len(dates))
print(len(messages))
print(df.shape)
```

```python
users = []
messages = []
for message in df['user_message']:
    entry = re.split(r'([\w\W]+?):\s', message)
    if entry[1:]:
        users.append(entry[1])
        messages.append(entry[2])
    else:
        users.append("group_notification")
        messages.append(entry[0])
df['user'] = users
df['message'] = messages
df.drop(columns=['user_message'], inplace=True)
```

```python
df.sample(20)
```

```python
# since we have converted it into pd date and time so we can directly use all these attributes 
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month_name()
df['day'] = df['date'].dt.day
df['hour'] = df['date'].dt.hour
df['minute'] = df['date'].dt.minute
df
```

```python
period = []
for hour in df[['day_name','hour']]['hour']:
    if hour == 23:
        period.append(str(hour) + "-" + str('00'))
    elif hour == 0:
        period.append(str('00') + "-" + str(hour+1))
    else:
        period.append(str(hour) + "-" + str(hour+1))
```

```python
df['period'] = period
```
